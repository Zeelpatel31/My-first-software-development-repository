```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Create the form fields.
title = widgets.HTML(
    value="<h2>Student Registration Form</h2>"
)

name_input = widgets.Text(
    description="Name:",
    placeholder="Enter your name"
)

email_input = widgets.Text(
    description="Email:",
    placeholder="Enter your email"
)

course_input = widgets.Dropdown(
    options=["Software Development", "Data Management",
             "Digital Marketing"],
    description="Course:"
)

register_button = widgets.Button(
    description="Register",
    button_style="success",
    icon="check"
)

output = widgets.Output()

# Process the registration when the button is clicked.
def register_student(button):
    with output:
        clear_output()

        name = name_input.value.strip()
        email = email_input.value.strip()
        course = course_input.value

        if not name or not email:
            print("Please complete all fields.")
        elif "@" not in email or "." not in email:
            print("Please enter a valid email address.")
        else:
            print("Registration Successful!")
            print("Student Name:", name)
            print("Email:", email)
            print("Selected Course:", course)

register_button.on_click(register_student)

# Display the complete GUI form.
display(
    title,
    name_input,
    email_input,
    course_input,
    register_button,
    output
)
```
