```python
# Storing multiple tasks in a list
tasks = [
    {
        "task_id": 1,
        "title": "Complete Python Assignment",
        "description": "Develop task management system",
        "status": "To Do",
        "priority": "High",
        "due_date": "2026-12-10"
    },
    {
        "task_id": 2,
        "title": "Prepare Project Documentation",
        "description": "Write project explanation and README",
        "status": "In Progress",
        "priority": "Medium",
        "due_date": "2026-12-15"
    },
    {
        "task_id": 3,
        "title": "Test Application",
        "description": "Check application features and errors",
        "status": "To Do",
        "priority": "Low",
        "due_date": "2026-12-20"
    }
]
print("Total tasks:", len(tasks))
for task in tasks:
    print(task["task_id"], "-", task["title"])
```

```python
# Function to display task details
def display_task(task):
    print("------------------------------")
    print("Task ID:", task["task_id"])
    print("Title:", task["title"])
    print("Description:", task["description"])
    print("Status:", task["status"])
    print("Priority:", task["priority"])
    print("Due Date:", task["due_date"])
    print("------------------------------")
# Calling the function
display_task(tasks[0])
```

```python
# Checking required task fields
required_fields = [
    "task_id",
    "title",
    "description",
    "status",
    "priority",
    "due_date"
]
for task in tasks:

    for field in required_fields:
        if field not in task:
            print("Missing field:", field)
print("Task structure validation completed!")
print("Number of sample tasks:", len(tasks))
print("All required task fields are available.")
```
