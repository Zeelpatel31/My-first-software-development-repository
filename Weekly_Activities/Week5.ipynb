```python
import random
import datetime

# Store quiz questions and their correct answers.
questions = [
    {
        "question": "Which language are we learning?",
        "options": ["1. Java", "2. Python", "3. C++"],
        "answer": "2"
    },
    {
        "question": "Which function displays output?",
        "options": ["1. print()", "2. input()", "3. type()"],
        "answer": "1"
    },
    {
        "question": "Which data type stores whole numbers?",
        "options": ["1. String", "2. Float", "3. Integer"],
        "answer": "3"
    }
]

# Shuffle the questions to make each game different.
random.shuffle(questions)

score = 0

print("WELCOME TO THE PYTHON QUIZ")
print("Date:", datetime.date.today())

# Display each question and check the user's answer.
for item in questions:
    print("\n" + item["question"])

    for option in item["options"]:
        print(option)

    answer = input("Enter your answer (1-3): ")

    if answer == item["answer"]:
        print("Correct answer!")
        score += 1
    else:
        print("Incorrect answer.")

# Display the final result.
print("\nQuiz completed!")
print("Your score:", score, "out of", len(questions))

if score == len(questions):
    print("Excellent performance!")
elif score >= 2:
    print("Good work!")
else:
    print("Keep practising!")
```
