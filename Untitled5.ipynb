````python
# Update Existing Task
def update_task():
    print("\n========== UPDATE TASK ==========")
    try:
        task_id = int(input("Enter Task ID to update: "))
    except ValueError:
        print("Error: Please enter a valid numeric Task ID.")
        return
    for task in tasks:
        if task["task_id"] == task_id:
            print("\nCurrent Task Details:")
            print(task)
            new_title = input("Enter new title (press Enter to keep current): ").strip()
            if new_title:
                task["title"] = new_title
            new_description = input("Enter new description (press Enter to keep current): ").strip()
            if new_description:
                task["description"] = new_description
            new_priority = input("Enter new priority (High/Medium/Low): ").strip().title()
            if new_priority:
                if new_priority not in ["High", "Medium", "Low"]:
                    print("Error: Invalid priority. Update cancelled.")
                    return
                task["priority"] = new_priority
            new_status = input("Enter new status (To Do/In Progress/Completed): ").strip().title()
            if new_status:
                if new_status not in ["To Do", "In Progress", "Completed"]:
                    print("Error: Invalid status. Update cancelled.")
                    return
                task["status"] = new_status
            new_due_date = input("Enter new due date (YYYY-MM-DD): ").strip()
            if new_due_date:
                try:
                    datetime.strptime(new_due_date, "%Y-%m-%d")
                except ValueError:
                    print("Error: Invalid date. Update cancelled.")
                    return
                task["due_date"] = new_due_date
            print("\nTask updated successfully!")
            print("Updated Task Details:")
            print(task)
            return
    print("Error: Task ID not found.")

update_task()
````

````python

import os
import json
from datetime import datetime

PROJECT_DIR = "/content/drive/MyDrive/CS73_Task_Management_System"
DATA_FILE = os.path.join(PROJECT_DIR, "tasks.json")

os.makedirs(PROJECT_DIR, exist_ok=True)

tasks = []
next_task_id = 1

PRIORITIES = ["High", "Medium", "Low"]
STATUSES = ["To Do", "In Progress", "Completed"]


def save_tasks():
    try:
        with open(DATA_FILE, "w") as file:
            json.dump(tasks, file, indent=4)
        print("Tasks saved successfully.")
    except (OSError, TypeError) as error:
        print("Error saving tasks:", error)


def load_tasks():
    global tasks, next_task_id

    if not os.path.exists(DATA_FILE):
        tasks = []
        next_task_id = 1
        return

    try:
        with open(DATA_FILE, "r") as file:
            loaded = json.load(file)

        if not isinstance(loaded, list):
            raise ValueError("Invalid task data format.")

        required = [
            "task_id", "title", "description",
            "status", "priority", "due_date"
        ]

        for task in loaded:
            if not isinstance(task, dict) or not all(
                field in task for field in required
            ):
                raise ValueError("A task has missing or invalid fields.")

        tasks = loaded
        next_task_id = max(
            (task["task_id"] for task in tasks),
            default=0
        ) + 1

    except (OSError, json.JSONDecodeError, ValueError, TypeError) as error:
        print("Error loading tasks:", error)
        tasks = []
        next_task_id = 1


def add_task():
    global next_task_id

    title = input("Enter task title: ").strip()
    if not title:
        print("Title cannot be empty.")
        return

    description = input("Enter description: ").strip()
    if not description:
        print("Description cannot be empty.")
        return

    priority = input("Priority (High/Medium/Low): ").strip().title()
    if priority not in PRIORITIES:
        print("Invalid priority.")
        return

    status = input(
        "Status (To Do/In Progress/Completed): "
    ).strip().title()

    if status not in STATUSES:
        print("Invalid status.")
        return

    due_date = input("Due date (YYYY-MM-DD): ").strip()

    try:
        datetime.strptime(due_date, "%Y-%m-%d")
    except ValueError:
        print("Invalid date. Use YYYY-MM-DD.")
        return

    task = {
        "task_id": next_task_id,
        "title": title,
        "description": description,
        "status": status,
        "priority": priority,
        "due_date": due_date
    }

    tasks.append(task)
    next_task_id += 1
    save_tasks()
    print("Task added successfully. ID:", task["task_id"])


def view_tasks():
    if not tasks:
        print("No tasks available.")
        return

    print("\n========== ALL TASKS ==========")

    for task in tasks:
        print("\nTask ID:", task["task_id"])
        print("Title:", task["title"])
        print("Description:", task["description"])
        print("Status:", task["status"])
        print("Priority:", task["priority"])
        print("Due Date:", task["due_date"])


def search_task():
    try:
        task_id = int(input("Enter Task ID: "))
    except ValueError:
        print("Please enter a numeric Task ID.")
        return

    for task in tasks:
        if task["task_id"] == task_id:
            print("\nTask Found:", task)
            return

    print("Task ID not found.")


def update_task():
    try:
        task_id = int(input("Enter Task ID to update: "))
    except ValueError:
        print("Please enter a numeric Task ID.")
        return

    task = next(
        (item for item in tasks if item["task_id"] == task_id),
        None
    )

    if task is None:
        print("Task ID not found.")
        return

    print("Press Enter to keep the existing value.")

    new_title = input("New title: ").strip()
    new_description = input("New description: ").strip()
    new_priority = input("New priority (High/Medium/Low): ").strip().title()
    new_status = input(
        "New status (To Do/In Progress/Completed): "
    ).strip().title()
    new_date = input("New due date (YYYY-MM-DD): ").strip()

    if new_priority and new_priority not in PRIORITIES:
        print("Invalid priority. No changes saved.")
        return

    if new_status and new_status not in STATUSES:
        print("Invalid status. No changes saved.")
        return

    if new_date:
        try:
            datetime.strptime(new_date, "%Y-%m-%d")
        except ValueError:
            print("Invalid date. No changes saved.")
            return

    if new_title:
        task["title"] = new_title
    if new_description:
        task["description"] = new_description
    if new_priority:
        task["priority"] = new_priority
    if new_status:
        task["status"] = new_status
    if new_date:
        task["due_date"] = new_date

    save_tasks()
    print("Task updated successfully.")


def complete_task():
    try:
        task_id = int(input("Enter Task ID to complete: "))
    except ValueError:
        print("Please enter a numeric Task ID.")
        return

    for task in tasks:
        if task["task_id"] == task_id:
            if task["status"] == "Completed":
                print("Task is already completed.")
                return

            task["status"] = "Completed"
            save_tasks()
            print("Task marked as completed.")
            return

    print("Task ID not found.")


def delete_task():
    try:
        task_id = int(input("Enter Task ID to delete: "))
    except ValueError:
        print("Please enter a numeric Task ID.")
        return

    for task in tasks:
        if task["task_id"] == task_id:
            confirmation = input(
                "Confirm deletion? (yes/no): "
            ).strip().lower()

            if confirmation == "yes":
                tasks.remove(task)
                save_tasks()
                print("Task deleted successfully.")
            else:
                print("Deletion cancelled.")
            return

    print("Task ID not found.")


def main_menu():
    load_tasks()

    while True:
        print("\n========== TASK MANAGEMENT SYSTEM ==========")
        print("1. Add New Task")
        print("2. View All Tasks")
        print("3. Search Task")
        print("4. Update Task")
        print("5. Mark Task as Completed")
        print("6. Delete Task")
        print("7. Exit")

        choice = input("Enter your choice (1-7): ").strip()

        if choice == "1":
            add_task()
        elif choice == "2":
            view_tasks()
        elif choice == "3":
            search_task()
        elif choice == "4":
            update_task()
        elif choice == "5":
            complete_task()
        elif choice == "6":
            delete_task()
        elif choice == "7":
            save_tasks()
            print("Thank you for using the application!")
            break
        else:
            print("Invalid choice. Select a number from 1 to 7.")


if __name__ == "__main__":
    main_menu()
````
