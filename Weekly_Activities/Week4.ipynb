```python
import csv
import os

inventory = {
    "Notebook": {"quantity": 20, "price": 45.50},
    "Pen": {"quantity": 50, "price": 10.00},
    "Backpack": {"quantity": 8, "price": 650.00}
}

def display_inventory():
    print("\n--- INVENTORY ---")
    for product, details in inventory.items():
        print(
            f"Product: {product} | "
            f"Quantity: {details['quantity']} | "
            f"Price: Rs.{details['price']:.2f}"
        )

def add_product():
    name = input("Enter product name: ").strip()

    if not name:
        print("Product name cannot be empty.")
        return

    if name in inventory:
        print("Product already exists.")
        return

    try:
        quantity = int(input("Enter quantity: "))
        price = float(input("Enter price: "))

        if quantity < 0 or price < 0:
            print("Quantity and price cannot be negative.")
            return

        inventory[name] = {"quantity": quantity, "price": price}
        print("Product added successfully.")

    except ValueError:
        print("Please enter valid numeric values.")

def update_product():
    name = input("Enter product name to update: ").strip()

    if name not in inventory:
        print("Product not found.")
        return

    try:
        quantity = int(input("Enter new quantity: "))
        price = float(input("Enter new price: "))

        if quantity < 0 or price < 0:
            print("Values cannot be negative.")
            return

        inventory[name]["quantity"] = quantity
        inventory[name]["price"] = price
        print("Product updated successfully.")

    except ValueError:
        print("Please enter valid numbers.")

def save_inventory():
    with open("inventory.csv", "w", newline="") as file:
        writer = csv.writer(file)
        writer.writerow(["Product", "Quantity", "Price"])

        for name, details in inventory.items():
            writer.writerow([
                name,
                details["quantity"],
                details["price"]
            ])

    print("Inventory saved to inventory.csv.")

def load_inventory():
    if not os.path.exists("inventory.csv"):
        print("No saved inventory file found.")
        return

    with open("inventory.csv", "r") as file:
        reader = csv.DictReader(file)
        inventory.clear()

        for row in reader:
            inventory[row["Product"]] = {
                "quantity": int(row["Quantity"]),
                "price": float(row["Price"])
            }

    print("Inventory loaded successfully.")

while True:
    print("\n1. Display Inventory")
    print("2. Add Product")
    print("3. Update Product")
    print("4. Save Inventory")
    print("5. Load Inventory")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        display_inventory()
    elif choice == "2":
        add_product()
    elif choice == "3":
        update_product()
    elif choice == "4":
        save_inventory()
    elif choice == "5":
        load_inventory()
    elif choice == "6":
        print("Inventory system closed.")
        break
    else:
        print("Invalid choice.")
```
